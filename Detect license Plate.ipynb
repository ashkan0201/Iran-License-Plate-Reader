```python
from ultralytics import YOLO
import cv2
import numpy as np
import os

model = YOLO("runs/segment/train/weights/best.pt")

img_path = "/content/drive/MyDrive/git hub/Iran-License-Plate-Reader/archive/test/images/288_png.rf.8e37dba6dcb939f6cd2565d5340b4934.jpg"

results = model.predict(
    img_path,
    conf=0.25,
    verbose=False,
    save=True
)

img = cv2.imread(img_path)

os.makedirs("mask_crop", exist_ok=True)

for r in results:
    if r.masks is None:
        continue

    for i, mask in enumerate(r.masks.data.cpu().numpy()):

        mask = cv2.resize(
            mask.astype(np.uint8),
            (img.shape[1], img.shape[0]),
            interpolation=cv2.INTER_NEAREST
        )

        # فقط ناحیه پلاک
        result = cv2.bitwise_and(img, img, mask=mask)

        # پیدا کردن محدوده واقعی ماسک
        ys, xs = np.where(mask > 0)

        if len(xs) == 0:
            continue

        x1, x2 = xs.min(), xs.max()
        y1, y2 = ys.min(), ys.max()

        # کراپ از تصویر ماسک شده
        plate_only = result[y1:y2, x1:x2]

        cv2.imwrite(
            f"mask_crop/plate_{i}.png",
            plate_only
        )

print("done")
```
