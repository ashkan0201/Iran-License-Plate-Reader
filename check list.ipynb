```python
# for removing 2 labels text file and image file
from pathlib import Path

mer_train = []
list_of_remove_item = []


def get_files(path):
    temp = []
    
    for x, y, z in os.walk(path):
        for i in z:
            temp.append(i)
        return temp

def delete_files(files, image_dir="archive/train/images", label_dir="archive/train/labels"):
    image_dir = Path(image_dir)
    label_dir = Path(label_dir)

    deleted = 0

    for item in files:
        img_path = image_dir / item["img"]
        label_path = label_dir / item["labels"]

        if img_path.exists():
            img_path.unlink()

        if label_path.exists():
            label_path.unlink()

        deleted += 1

    print(f"{deleted} samples removed.")

def checked_for_removeing_labels_file(path):
    temp_5 = 0
    temp_11 = 0
    for i in mer_train:
        with open(f"{path}/{i["labels"]}", "r") as file:
            data = file.readline()

            if len(data.split(" ")) == 5:
                temp_5 += 1

            else:
                temp_11 += 1
    print(f"({temp_11}, {temp_5})")

list_of_train_img = get_files("archive/train/images/")
list_of_train_labels = get_files("archive/train/labels/")

for i in list_of_train_img:
    for j in list_of_train_labels:
        if i[:-4] == j[:-4]:
            mer_train.append({"img":i, "labels":j})


for i in mer_train:
    with open(f"archive/train/labels/{i["labels"]}", "r") as file:
        data = file.readline()
        
        if len(data.split(" ")) == 5:
            list_of_remove_item.append(i)
```
